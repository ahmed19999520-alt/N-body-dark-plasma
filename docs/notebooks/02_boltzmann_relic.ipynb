# Coupled Boltzmann hierarchy and relic abundance

```
from N_bodyDP.boltzmann import CoannihilationModel, BoltzmannSolver, calibrate_to_relic, effective_dof, solve_freeze_out, sigma_eff
from N_bodyDP.constants import GEV_INV2_TO_CM3_S
raw = CoannihilationModel()
model = raw.with_target_sigma_eff(3.0e-26, 27.0)
effective_dof(model, 27.0), solve_freeze_out(model), sigma_eff(model, 27.0) * GEV_INV2_TO_CM3_S
```

```
calibrated = calibrate_to_relic(model)
result = BoltzmannSolver(calibrated).solve(n_eval=400)
result.omega_h2, result.y_inf, result.success
```

```
from N_bodyDP.plotting import boltzmann_yield_plot
boltzmann_yield_plot(result)
```
