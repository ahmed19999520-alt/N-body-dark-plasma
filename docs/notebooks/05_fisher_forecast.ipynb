# Fisher forecast and Bayesian reconstruction

```
from N_bodyDP.gravitational_waves import ConfinementTransition, LisaDetector
det = LisaDetector()
tr = ConfinementTransition()
det.forecast(tr)
```

```
from N_bodyDP.plotting import corner_plot
corner_plot(result, truth)
```
